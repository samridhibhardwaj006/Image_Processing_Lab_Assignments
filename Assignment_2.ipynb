```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Read RGB image
img = Image.open("doggy.jpeg").convert("RGB")
img = np.array(img)

print("Image loaded successfully!")
print("Image size:", img.shape)



gray_mean = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):

        # Convert to int to avoid overflow
        R = int(img[i, j, 0])
        G = int(img[i, j, 1])
        B = int(img[i, j, 2])

        gray_mean[i, j] = int((R + G + B) / 3)

print("Case 1 completed!")




print("\nEnter weightages for R, G and B.")
print("Their sum must be equal to 1.")

R_weight = float(input("Enter weightage of R: "))
G_weight = float(input("Enter weightage of G: "))
B_weight = float(input("Enter weightage of B: "))

if (0 <= R_weight <= 1 and
    0 <= G_weight <= 1 and
    0 <= B_weight <= 1 and
    abs(R_weight + G_weight + B_weight - 1) < 0.0001):

    gray_weighted = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):

            R = int(img[i, j, 0])
            G = int(img[i, j, 1])
            B = int(img[i, j, 2])

            gray_weighted[i, j] = int(
                R_weight * R +
                G_weight * G +
                B_weight * B
            )

    print("Case 2 completed!")



    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img)
    plt.title("Original RGB Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(gray_mean, cmap="gray")
    plt.title("Mean Average")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(gray_weighted, cmap="gray")
    plt.title("Weighted Average")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

else:
    print("Invalid weightages!")
    print("Each weight must be between 0 and 1.")
    print("Their sum must be equal to 1.")
```
