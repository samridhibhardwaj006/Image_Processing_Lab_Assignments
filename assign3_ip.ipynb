```python
import cv2
import numpy as np
import matplotlib.pyplot as plt



filename = "doggy.jpeg"
img_gray = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)


if img_gray is None:
    print("Image not found! Make sure doggy.jpeg is uploaded to Colab.")
else:
    print("Image loaded successfully!")
    print("Image size:", img_gray.shape)


    # Q1


    border_width = int(input("Enter border width (pixels): "))
    border_color = int(input("Enter grayscale border color (0-255): "))


    bordered_gray = cv2.copyMakeBorder(
        img_gray,
        border_width, border_width,
        border_width, border_width,
        cv2.BORDER_CONSTANT,
        value=border_color
    )


    _, img_binary = cv2.threshold(
        img_gray, 127, 255, cv2.THRESH_BINARY
    )

    binary_border_color = int(
        input("Enter binary border color (0 or 255): ")
    )

    bordered_binary = cv2.copyMakeBorder(
        img_binary,
        border_width, border_width,
        border_width, border_width,
        cv2.BORDER_CONSTANT,
        value=binary_border_color
    )

   #Q2

    complement_gray = 255 - img_gray
    complement_binary = 255 - img_binary



    plt.figure(figsize=(12, 8))

    plt.subplot(2, 3, 1)
    plt.imshow(img_gray, cmap="gray")
    plt.title("Original Grayscale")
    plt.axis("off")

    plt.subplot(2, 3, 2)
    plt.imshow(bordered_gray, cmap="gray")
    plt.title("Grayscale + Border")
    plt.axis("off")

    plt.subplot(2, 3, 3)
    plt.imshow(complement_gray, cmap="gray")
    plt.title("Grayscale Complement")
    plt.axis("off")

    plt.subplot(2, 3, 4)
    plt.imshow(img_binary, cmap="gray")
    plt.title("Binary Image")
    plt.axis("off")

    plt.subplot(2, 3, 5)
    plt.imshow(bordered_binary, cmap="gray")
    plt.title("Binary + Border")
    plt.axis("off")

    plt.subplot(2, 3, 6)
    plt.imshow(complement_binary, cmap="gray")
    plt.title("Binary Complement")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
