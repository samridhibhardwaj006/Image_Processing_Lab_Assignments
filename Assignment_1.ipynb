```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = Image.open("dog.jpeg").convert("L")
img = np.array(img)

print("Image loaded successfully.")
print("Image size:", img.shape)



mean_threshold = np.mean(img)

binary_mean = np.zeros_like(img)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        if img[i, j] >= mean_threshold:
            binary_mean[i, j] = 255
        else:
            binary_mean[i, j] = 0

print("Mean intensity threshold:", mean_threshold)




threshold = int(input("Enter threshold value (0-255): "))

if threshold < 0 or threshold > 255:
    print("Invalid threshold! Enter a value between 0 and 255.")

else:
    binary_user = np.zeros_like(img)

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if img[i, j] >= threshold:
                binary_user[i, j] = 255
            else:
                binary_user[i, j] = 0


    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img, cmap="gray")
    plt.title("Original Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(binary_mean, cmap="gray")
    plt.title("Mean Threshold")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(binary_user, cmap="gray")
    plt.title("User Threshold")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
